```python
import numpy as np
```

活性化関数: 入力信号の総和を出力信号に変換する関数

```python
def simple_step_function(x):
  """
  ステップ関数: 0超過で1、それ以外で0を返す
  """
  if x > 0:
    return 1
  else:
    return 0

# Numpy配列に対応した書き方
def step_function(x):
  y = x > 0                   # 配列の各要素をブール値に変換
  return y.astype(int)     # 配列の各ブール値を0か1に変換

def sigmoid(x):
  """
  シグモイド関数: 0～1の滑らかな値を返す
  """
  return 1 / (1 + np.exp(-x))

def ReLU(x):
  """
  ReLU(Rectified Linear Unit: 整流された線形ユニット)関数: 0超過でそのまま、それ以外で0を返す
  """
  return np.maximum(0, x)     # 2つの配列に関して一つずつ比較して大きい方を返す(ブロードキャスト対応)

def softmax(a):               # ソフトマックス関数は出力層で使うので a を使用している
  """
  ソフトマックス関数: 総和が1となる出力信号を返す(確率であると解釈可能)
  """
  c = np.max(a)
  exp_a = np.exp(a - c)       # オーバーフロー対策
  sum_exp_a = np.sum(exp_a)

  y = exp_a / sum_exp_a
  return y

def Tanh(x):
  """
  Tanh(ハイパボリックタンジェント): -1～1の滑らかな値を返す
  """
  return np.tanh(x)
```

```python
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)

y1 = step_function(x)
y2 = sigmoid(x)
y3 = ReLU(x)
y4 = Tanh(x)

plt.figure(figsize=(10,5))

plt.subplot(2,2,1)
plt.ylabel("Step")
plt.plot(x, y1, color="blue")
plt.grid(True)

plt.subplot(2,2,2)
plt.ylabel("Sigmoid")
plt.plot(x, y2, color="green")
plt.grid(True)

plt.subplot(2,2,3)
plt.ylabel("ReLU")
plt.plot(x, y3, color="red")
plt.grid(True)

plt.subplot(2,2,4)
plt.ylabel("Tanh")
plt.plot(x, y4, color="orange")
plt.grid(True)

plt.tight_layout()

plt.show()
```
