```python
import numpy as np

"""
活性化関数
"""
def step_function(x):
  y = x > 0
  return y.astype(int)

def sigmoid(x):
  return 1 / (1 + np.exp(-x))

def ReLU(x):
  return np.maximum(0, x)

def softmax(a):
  c = np.max(a)
  exp_a = np.exp(a - c)
  sum_exp_a = np.sum(exp_a)
  y = exp_a / sum_exp_a
  return y

def Tanh(x):
  return np.tanh(x)
```

```python
def init_network():
  """
  重みとバイアスの初期化
  """
  network = {}
  network['W1'] = np.array([[0.1,0.3,0.5],[0.2,0.4,0.6]])
  network['b1'] = np.array([0.1,0.2,0.3])
  network['W2'] = np.array([[0.1,0.4],[0.2,0.5],[0.3,0.6]])
  network['b2'] = np.array([0.1,0.2])
  network['W3'] = np.array([[10.0, 10.0], [10.0, 10.0]])
  network['b3'] = np.array([-5.0, -5.0])

  return network

def forward(network, x):
  """
  3層ニューラルネットワークの順伝播
  """
  W1, W2, W3 = network['W1'], network['W2'], network['W3']
  b1, b2, b3 = network['b1'], network['b2'], network['b3']

  a1 = np.dot(x, W1) + b1
  z1 = sigmoid(a1)
  a2 = np.dot(z1, W2) + b2
  z2 = sigmoid(a2)
  a3 = np.dot(z2, W3) + b3

  y = a3       # 恒等関数
  return y
```

```python
network = init_network()
x = np.array([1.0, 0.5])
print(forward(network, x))
```

```python
import pickle


def get_data():
  (x_train, t_train), (x_test, t_test) = load_mnist(flatten=True, normalize=True, one_hot_label=False)
  return x_test, t_test

def init_network():
  with open("/content/drive/MyDrive/Colab Notebooks/sample_weight.pkl", "rb") as f:
    network = pickle.load(f)

  return network

network = init_network()

# 予測
x_test, t_test = get_data()
accuracy_cnt = 0

def forward(network, x):
  """
  3層ニューラルネットワークの順伝播
  """
  W1, W2, W3 = network['W1'], network['W2'], network['W3']
  b1, b2, b3 = network['b1'], network['b2'], network['b3']

  a1 = np.dot(x, W1) + b1
  z1 = sigmoid(a1)
  a2 = np.dot(z1, W2) + b2
  z2 = sigmoid(a2)
  a3 = np.dot(z2, W3) + b3

  y = softmax(a3)
  return y

# テストデータごとに予測＆評価
for i in range(len(x_test)):
  y = forward(network, x_test[i])
  p = np.argmax(y)

  if p == t_test[i]:
    accuracy_cnt += 1

print(f"正解率: {accuracy_cnt / len(x_test)}")
```
