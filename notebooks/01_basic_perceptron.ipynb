```python
import numpy as np
```

```python
"""
シンプルなパーセプトロン
"""

# AND
def AND(x1,x2):
  x = np.array([x1, x2])      # 入力信号(Input Signal)
  w = np.array([0.5, 0.5])    # 重み(Weight)
  b = -0.7                    # バイアス(Bias)

  tmp = np.sum(x * w) + b
  if tmp <= 0:
    return 0
  else:
    return 1

print(AND(1,1), AND(1,0), AND(0,1), AND(0,0))

# OR
def OR(x1,x2):
  x = np.array([x1, x2])
  w = np.array([0.5, 0.5])
  b = -0.4

  tmp = np.sum(x * w) + b
  if tmp <= 0:
    return 0
  else:
    return 1

print(OR(1,1), OR(1,0), OR(0,1), OR(0,0))
```
